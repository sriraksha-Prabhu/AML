```python
#19

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],
    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],
    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],
    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],
    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)

# 1. Display first five records
print(df.head())

# 2. Find dimensions of the dataset
print(df.shape)

# 3. Find average final mark
print(df["FinalMarks"].mean())

# 4. Find highest-scoring student
print(df.loc[df["FinalMarks"].idxmax()])

# 5. Find lowest-scoring student
print(df.loc[df["FinalMarks"].idxmin()])

# 6. Display students scoring above 80
print(df[df["FinalMarks"] > 80])

# 7. Find average marks by department
print(df.groupby("Department")["FinalMarks"].mean())

# 8. Students having attendance above 80%
print(df[df["Attendance"] > 80])

# 9. Sort students by FinalMarks
print(df.sort_values("FinalMarks"))

# 10. Add Result column
df["Result"] = df["FinalMarks"].apply(
    lambda x: "Pass" if x >= 70 else "Fail"
)

# 11. Add Performance Category
df["Performance"] = df["FinalMarks"].apply(
    lambda x: "Excellent" if x >= 90
    else "Very Good" if x >= 80
    else "Good" if x >= 70
    else "Needs Improvement"
)

print(df)

# 12. Department having highest average FinalMarks
average = df.groupby("Department")["FinalMarks"].mean()
print(average.idxmax())
```

```python
#20
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],
    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],
    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

# 1. Average salary
print(employees["Salary"].mean())

# 2. Maximum salary
print(employees["Salary"].max())

# Minimum salary
print(employees["Salary"].min())

# 3. Employee receiving the highest salary
print(employees[employees["Salary"] == employees["Salary"].max()])

# 4. Employees earning above ₹60,000
print(employees[employees["Salary"] > 60000])

# 5. Average salary for each department
print(employees.groupby("Department")["Salary"].mean())

# 6. Average experience by department
print(employees.groupby("Department")["Experience"].mean())

# 7. Number of employees in each department
print(employees.groupby("Department")["Employee"].count())

# 8. Sort employees by salary
print(employees.sort_values("Salary"))

# 9. IT employees with more than 3 years of experience
print(employees[
    (employees["Department"] == "IT") &
    (employees["Experience"] > 3)
])

# 10. Add SalaryLevel column
employees["SalaryLevel"] = employees["Salary"].apply(
    lambda x: "High" if x >= 75000
    else "Medium" if x >= 50000
    else "Low"
)

print(employees)
```
