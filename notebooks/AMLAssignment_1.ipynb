```python
#12
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])
mean_temp = np.mean(temperature)
max_temp = np.max(temperature)
min_temp = np.min(temperature)
above_average = np.sum(temperature > mean_temp)
greater_34 = temperature[temperature > 34]
difference = max_temp - min_temp
std_dev = np.std(temperature)
print("Mean temperature:", mean_temp)
print("Maximum temperature:", max_temp)
print("Minimum temperature:", min_temp)
print("Days above average:", above_average)
print("Temperatures greater than 34°C:", greater_34)
print("Difference:", difference)
print("Standard deviation:", std_dev)
```

```python
#13
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])
print("Total annual sales",sales.sum())
print("Average monthly sales",sales.mean())
print("Month with the highest sales",np.argmax(sales)+1)
print("Month with the lowest sales",np.argmin(sales)+1)
print("Number of months with sales above ₹20,000",np.sum(sales>20000))
print("Percentage increase from January to December",(sales[-1]-sales[0])/sales[0]*100)
```

```python
#14 Consider marks of five students in four subjects:
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])
print("Total marks of every student",np.sum(marks,axis=1))
print("Average marks of every student",np.mean(marks,axis=1))
print("Average marks of every subject",np.mean(marks,axis=0))
print("Highest-scoring student",np.argmax(np.sum(marks,axis=1)))
print("Lowest-scoring student",np.argmin(np.sum(marks,axis=1)))
print("Highest-scoring subject",np.argmax(np.mean(marks,axis=0)))
print("Students whose average is greater than 80",np.sum(np.mean(marks,axis=1)>80))
print("Overall class average",np.mean(np.mean(marks,axis=1)))
```

```python
student_average = np.mean(marks, axis=1)
conditions = [
    student_average >= 85,
    student_average >= 70,
    student_average >= 50
]

categories = [
    "Excellent",
    "Good",
    "Average"
]

performance = np.select(
    conditions,
    categories,
    default="Needs Improvement"
)
print("\nPerformance category:")
for i in range(len(performance)):
    print("Student", i + 1, ":", performance[i])
```

```python
#23
marks = np.random.randint(0, 101, 100)

print("Exam marks:")
print(marks)


range_0_39 = np.sum((marks >= 0) & (marks <= 39))
range_40_59 = np.sum((marks >= 40) & (marks <= 59))
range_60_79 = np.sum((marks >= 60) & (marks <= 79))
range_80_100 = np.sum((marks >= 80) & (marks <= 100))

print("\nNumber of students:")
print("0–39   :", range_0_39)
print("40–59  :", range_40_59)
print("60–79  :", range_60_79)
print("80–100 :", range_80_100)
```

```python

import time


numbers = list(range(1000000))
array = np.arange(1000000)


start = time.time()

list_result = [x * 2 for x in numbers]

list_time = time.time() - start


start = time.time()

array_result = array * 2

numpy_time = time.time() - start

print("Python list time :", list_time, "seconds")
print("NumPy array time :", numpy_time, "seconds")
```
